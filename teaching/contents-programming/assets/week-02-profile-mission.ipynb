# WEEK 02 MISSION: 나의 콘텐츠 생활을 데이터로 표현하기

아래 코드 셀의 오른쪽 값을 자신의 정보로 바꾸고, 계산하고, 출력한 뒤 마지막 자동 검사 셀을 실행합니다. 자동 검사에서 `WEEK 02 MISSION COMPLETE`가 나오면 `.ipynb` 파일을 제출할 수 있습니다.

주소, 전화번호, 생년월일처럼 공개하고 싶지 않은 정보는 작성하지 않습니다. 이름 대신 별명을 사용해도 됩니다.

```python
# STEP 1: = 오른쪽의 값만 자신의 정보로 바꾸세요.
student_name = "여기에 작성"
major = "여기에 작성"
favorite_content = "여기에 작성"
weekly_hours = -1
average_rating = -1.0
wants_to_create = "True 또는 False로 바꾸세요"

# 아래 실행 검사용 코드는 수정하지 마세요.
mission_step1_execution = get_ipython().execution_count
mission_stage = 1
```

```python
# STEP 2: 일주일 값을 이용해 한 달 예상 시간을 계산하세요.
assert mission_stage == 1, "STEP 1부터 다시 실행하세요."
mission_step2_execution = get_ipython().execution_count
monthly_hours = weekly_hours * 4
mission_stage = 2
```

```python
# STEP 3: 저장한 값과 계산 결과를 출력하세요.
assert mission_stage == 2, "STEP 2 계산 셀을 먼저 실행하세요."
mission_step3_execution = get_ipython().execution_count
print("이름 또는 별명:", student_name)
print("전공:", major)
print("좋아하는 콘텐츠:", favorite_content)
print("일주일 감상 시간:", weekly_hours, "시간")
print("콘텐츠 평점:", average_rating)
print("직접 만들고 싶은가:", wants_to_create)
print("한 달 예상 감상 시간:", monthly_hours, "시간")

initial_output_snapshot = (student_name, major, favorite_content, weekly_hours, average_rating, wants_to_create, monthly_hours)
mission_stage = 3
```

## STEP 4: 값 변경 실험

STEP 1부터 STEP 3까지 실행해 첫 출력을 확인합니다. 아래 셀의 `weekly_hours`와 `average_rating`에는 처음과 다른 값을 작성합니다. 이 셀은 같은 변수에 새 값을 다시 대입하고 변경 전후를 함께 출력합니다.

```python
# STEP 4: 오른쪽의 두 임시 값을 처음과 다른 값으로 바꾸세요.
assert mission_stage == 3, "STEP 3 출력 셀을 먼저 실행하세요."
mission_step4_execution = get_ipython().execution_count
initial_weekly_hours = weekly_hours
initial_average_rating = average_rating
weekly_hours = -2
average_rating = -2.0
monthly_hours = weekly_hours * 4

print("변경 전 일주일 감상 시간:", initial_weekly_hours, "시간")
print("변경 후 일주일 감상 시간:", weekly_hours, "시간")
print("변경 전 콘텐츠 평점:", initial_average_rating)
print("변경 후 콘텐츠 평점:", average_rating)
print("변경 후 한 달 예상 감상 시간:", monthly_hours, "시간")

changed_output_snapshot = (initial_weekly_hours, weekly_hours, initial_average_rating, average_rating, monthly_hours)
mission_stage = 4
```
