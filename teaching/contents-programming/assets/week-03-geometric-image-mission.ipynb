# WEEK 03 MISSION: 좌표와 RGB로 기하학적 이미지 완성하기

제공된 기준 이미지를 먼저 실행한 뒤, **색상 역할 2개 이상**과 **도형 좌표 묶음 2개 이상**을 수정해 자신의 기하학적 이미지를 완성합니다. 마지막 자동 검사에서 `WEEK 03 IMAGE MISSION COMPLETE`가 나오고 `.ipynb`와 `.png` 두 파일을 제출하면 남은 시간과 관계없이 퇴실할 수 있습니다.

빈 화면에서 전체 코드를 외워 작성하지 않습니다. `EDIT`라고 표시된 값만 바꾸고, `DO NOT EDIT` 영역은 그대로 둡니다. 한 번에 값 하나를 바꾼 뒤 다시 실행해 결과를 비교합니다.

```python
# STEP 0: 준비 셀입니다. 이 셀은 수정하지 말고 실행만 하세요.
from pathlib import Path
from PIL import Image, ImageDraw
from IPython.display import display

canvas_width = 600
canvas_height = 400

STARTER_BACKGROUND_COLOR = (244, 239, 229)
STARTER_PRIMARY_COLOR = (235, 86, 75)
STARTER_SECONDARY_COLOR = (48, 95, 180)
STARTER_ACCENT_COLOR = (244, 190, 55)
STARTER_OUTLINE_COLOR = (28, 31, 38)
STARTER_RECTANGLE_BOX = (50, 60, 330, 240)
STARTER_ELLIPSE_BOX = (230, 100, 510, 360)
STARTER_TRIANGLE_POINTS = ((390, 55), (560, 220), (340, 220))

mission_step0_execution = get_ipython().execution_count
mission_stage = 0
print("✅ STEP 0 준비 완료: 다음 셀로 이동하세요.")
```

## STEP 1: 제출 정보와 표현 의도 작성

`=` 오른쪽의 네 문자열만 수정합니다. `design_intent`에는 어떤 도형을 가장 먼저 보이게 할지와 그 이유를 한 문장으로 작성합니다. 파일명은 `week03_학번_이름.png` 형식을 유지합니다.

```python
# STEP 1 / EDIT: 따옴표 안의 내용만 자신의 정보로 바꾸세요.
assert mission_stage in (0, 1, 2, 3, 4), "STEP 0 준비 셀을 먼저 실행하세요."
student_id = "학번"
student_name = "이름 또는 별명"
design_intent = "가장 먼저 보이게 할 도형과 그 이유를 한 문장으로 작성하세요."
output_filename = "week03_학번_이름.png"

# DO NOT EDIT: 아래 실행 확인 코드는 수정하지 마세요.
mission_step1_execution = get_ipython().execution_count
mission_stage = 1
print("표현 의도:", design_intent)
print("저장 파일명:", output_filename)
```

## STEP 2: 역할에 따라 팔레트 수정

처음에는 기준값 그대로 실행해 다섯 색을 확인합니다. 그다음 `background_color`, `primary_color`, `secondary_color`, `accent_color`, `outline_color` 가운데 **최소 두 줄의 RGB 숫자**를 바꿉니다. 각 숫자는 0부터 255 사이의 정수로 작성합니다.

```python
# STEP 2 / EDIT: 아래 다섯 RGB 튜플의 숫자만 수정하세요.
assert mission_stage in (1, 2, 3, 4), "STEP 1 제출 정보 셀을 먼저 실행하세요."
background_color = (244, 239, 229)
primary_color = (235, 86, 75)
secondary_color = (48, 95, 180)
accent_color = (244, 190, 55)
outline_color = (28, 31, 38)

# DO NOT EDIT: 아래 코드는 RGB 범위와 팔레트 미리보기를 만듭니다.
current_palette = (background_color, primary_color, secondary_color, accent_color, outline_color)
assert all(type(color) is tuple and len(color) == 3 for color in current_palette), "모든 색을 (R, G, B) 세 정수로 작성하세요."
assert all(type(channel) is int and 0 <= channel <= 255 for color in current_palette for channel in color), "RGB 값은 따옴표 없는 0~255 정수여야 합니다."

palette_preview = Image.new("RGB", (500, 100), background_color)
palette_draw = ImageDraw.Draw(palette_preview)
palette_draw.rectangle((100, 0, 199, 99), fill=primary_color)
palette_draw.rectangle((200, 0, 299, 99), fill=secondary_color)
palette_draw.rectangle((300, 0, 399, 99), fill=accent_color)
palette_draw.rectangle((400, 0, 499, 99), fill=outline_color)
display(palette_preview)

starter_palette = (STARTER_BACKGROUND_COLOR, STARTER_PRIMARY_COLOR, STARTER_SECONDARY_COLOR, STARTER_ACCENT_COLOR, STARTER_OUTLINE_COLOR)
changed_palette_count = sum(current != starter for current, starter in zip(current_palette, starter_palette))
print("현재 변경한 색상 역할:", changed_palette_count, "개 / 완료 조건은 2개 이상")
mission_step2_execution = get_ipython().execution_count
mission_stage = 2
```

## STEP 3: 도형 좌표 수정과 이미지 생성

처음에는 기준 좌표 그대로 실행해 이미지를 확인합니다. 그다음 사각형, 타원, 삼각형 가운데 **최소 두 도형의 좌표 묶음**을 수정합니다. 사각형과 타원은 `(left, top, right, bottom)`, 삼각형은 세 개의 `(x, y)` 점을 사용합니다.

수정할 때에는 숫자 하나만 바꾸고 다시 실행합니다. 빨간 오류가 나오면 왼쪽·위 좌표가 오른쪽·아래 좌표보다 작은지, 모든 점이 600 × 400 캔버스 안에 있는지 확인합니다.

```python
# STEP 3 / EDIT: 아래 세 좌표 묶음의 숫자만 수정하세요.
assert mission_stage in (2, 3, 4), "STEP 2 팔레트 셀을 먼저 실행하세요."
rectangle_box = (50, 60, 330, 240)
ellipse_box = (230, 100, 510, 360)
triangle_points = (
    (390, 55),
    (560, 220),
    (340, 220)
)

# DO NOT EDIT: 아래 코드는 좌표를 확인하고 최종 이미지를 그립니다.
assert type(rectangle_box) is tuple and len(rectangle_box) == 4 and all(type(value) is int for value in rectangle_box), "rectangle_box는 정수 네 개의 튜플이어야 합니다."
assert type(ellipse_box) is tuple and len(ellipse_box) == 4 and all(type(value) is int for value in ellipse_box), "ellipse_box는 정수 네 개의 튜플이어야 합니다."
assert type(triangle_points) is tuple and len(triangle_points) == 3, "triangle_points에는 점 세 개가 필요합니다."
assert all(type(point) is tuple and len(point) == 2 and all(type(value) is int for value in point) for point in triangle_points), "삼각형의 각 점을 (x, y) 정수 튜플로 작성하세요."

rectangle_left, rectangle_top, rectangle_right, rectangle_bottom = rectangle_box
ellipse_left, ellipse_top, ellipse_right, ellipse_bottom = ellipse_box
assert 0 <= rectangle_left < rectangle_right < canvas_width and 0 <= rectangle_top < rectangle_bottom < canvas_height, "사각형 좌표를 캔버스 안에서 왼쪽 < 오른쪽, 위 < 아래 순서로 작성하세요."
assert 0 <= ellipse_left < ellipse_right < canvas_width and 0 <= ellipse_top < ellipse_bottom < canvas_height, "타원 좌표를 캔버스 안에서 왼쪽 < 오른쪽, 위 < 아래 순서로 작성하세요."
assert rectangle_right - rectangle_left >= 60 and rectangle_bottom - rectangle_top >= 60, "사각형의 가로와 세로를 각각 60픽셀 이상으로 만드세요."
assert ellipse_right - ellipse_left >= 60 and ellipse_bottom - ellipse_top >= 60, "타원의 가로와 세로를 각각 60픽셀 이상으로 만드세요."
assert all(0 <= x < canvas_width and 0 <= y < canvas_height for x, y in triangle_points), "삼각형의 모든 점을 600 × 400 캔버스 안에 두세요."
triangle_a, triangle_b, triangle_c = triangle_points
triangle_area_twice = abs(triangle_a[0] * (triangle_b[1] - triangle_c[1]) + triangle_b[0] * (triangle_c[1] - triangle_a[1]) + triangle_c[0] * (triangle_a[1] - triangle_b[1]))
assert triangle_area_twice >= 2000, "삼각형의 세 점이 한 줄에 놓이지 않도록 더 넓게 배치하세요."

image = Image.new("RGB", (canvas_width, canvas_height), background_color)
draw = ImageDraw.Draw(image)
draw.rectangle(rectangle_box, fill=primary_color, outline=outline_color, width=4)
draw.ellipse(ellipse_box, fill=secondary_color, outline=outline_color, width=4)
draw.polygon(triangle_points, fill=accent_color, outline=outline_color, width=4)
draw.line(((40, 340), (560, 340)), fill=outline_color, width=6)
display(image)

starter_geometry = (STARTER_RECTANGLE_BOX, STARTER_ELLIPSE_BOX, STARTER_TRIANGLE_POINTS)
current_geometry = (rectangle_box, ellipse_box, triangle_points)
changed_geometry_count = sum(current != starter for current, starter in zip(current_geometry, starter_geometry))
print("현재 변경한 도형 좌표 묶음:", changed_geometry_count, "개 / 완료 조건은 2개 이상")
mission_step3_execution = get_ipython().execution_count
mission_stage = 3
```

## STEP 4: PNG 저장

표현 의도와 맞는 최종 이미지를 확인한 뒤 아래 셀을 실행합니다. 저장 후 표시되는 이미지와 파일명을 확인합니다. 색상이나 좌표를 다시 바꾸었다면 STEP 2 또는 STEP 3부터 다시 실행한 뒤 이 저장 셀을 다시 실행합니다.

```python
# STEP 4: 최종 이미지를 PNG로 저장합니다. 이 셀은 수정하지 마세요.
assert mission_stage == 3, "STEP 3 이미지 생성 셀을 먼저 실행하세요."
mission_step4_execution = get_ipython().execution_count
image.save(output_filename)
with Image.open(output_filename) as opened_image:
    saved_image = opened_image.copy()
display(saved_image)
print("✅ PNG 저장 완료:", output_filename)
mission_stage = 4
```

```python
# FINAL CHECK: 이 셀의 코드는 수정하지 마세요.
mission_final_execution = get_ipython().execution_count

assert type(student_id) is str and student_id.strip() not in ("", "학번"), "STEP 1의 student_id를 자신의 학번으로 바꾸세요."
assert type(student_name) is str and student_name.strip() not in ("", "이름 또는 별명"), "STEP 1의 student_name을 이름 또는 별명으로 바꾸세요."
assert type(design_intent) is str and len(design_intent.strip()) >= 10 and design_intent != "가장 먼저 보이게 할 도형과 그 이유를 한 문장으로 작성하세요.", "design_intent에 표현 의도를 10자 이상 작성하세요."
expected_output_filename = "week03_" + student_id + "_" + student_name + ".png"
assert output_filename == expected_output_filename, "output_filename을 " + expected_output_filename + "으로 작성하세요."

current_palette = (background_color, primary_color, secondary_color, accent_color, outline_color)
starter_palette = (STARTER_BACKGROUND_COLOR, STARTER_PRIMARY_COLOR, STARTER_SECONDARY_COLOR, STARTER_ACCENT_COLOR, STARTER_OUTLINE_COLOR)
assert all(type(color) is tuple and len(color) == 3 for color in current_palette), "모든 색을 (R, G, B) 튜플로 작성하세요."
assert all(type(channel) is int and 0 <= channel <= 255 for color in current_palette for channel in color), "모든 RGB 채널은 0~255 정수여야 합니다."
assert len(set(current_palette)) == 5, "배경·주조·보조·강조·윤곽에 서로 다른 색을 사용하세요."
changed_palette_count = sum(current != starter for current, starter in zip(current_palette, starter_palette))
assert changed_palette_count >= 2, "기준 팔레트에서 색상 역할을 최소 두 개 변경하세요."

current_geometry = (rectangle_box, ellipse_box, triangle_points)
starter_geometry = (STARTER_RECTANGLE_BOX, STARTER_ELLIPSE_BOX, STARTER_TRIANGLE_POINTS)
changed_geometry_count = sum(current != starter for current, starter in zip(current_geometry, starter_geometry))
assert changed_geometry_count >= 2, "사각형·타원·삼각형 중 최소 두 도형의 좌표 묶음을 변경하세요."
assert isinstance(image, Image.Image) and image.mode == "RGB" and image.size == (600, 400), "600 × 400 RGB 이미지를 다시 생성하세요."
visible_color_counts = image.getcolors(maxcolors=canvas_width * canvas_height)
assert visible_color_counts is not None, "이미지의 색상 정보를 확인할 수 없습니다."
visible_colors = {color for count, color in visible_color_counts}
assert all(color in visible_colors for color in (background_color, primary_color, secondary_color, accent_color, outline_color)), "배경·사각형·타원·삼각형·윤곽색이 모두 보이도록 좌표와 겹침을 조정하세요."

assert Path(output_filename).is_file(), "STEP 4 저장 셀을 실행해 PNG 파일을 만드세요."
with Image.open(output_filename) as checked_file:
    assert checked_file.format == "PNG", "파일 확장자와 저장 형식을 PNG로 사용하세요."
    assert checked_file.mode == "RGB" and checked_file.size == image.size, "저장된 PNG의 모드와 크기를 확인하세요."
    assert checked_file.tobytes() == image.tobytes(), "이미지를 수정한 뒤 STEP 4 저장 셀을 다시 실행하세요."

assert mission_stage == 4, "STEP 0부터 STEP 4까지 순서대로 실행하세요."
assert (mission_step0_execution, mission_step1_execution, mission_step2_execution, mission_step3_execution, mission_step4_execution, mission_final_execution) == (1, 2, 3, 4, 5, 6), "세션을 다시 시작한 뒤 위에서 아래로 모두 실행하세요."

print("✅ 제출 정보와 표현 의도 검사 통과")
print("✅ RGB 범위와 팔레트 변경 검사 통과")
print("✅ 세 도형과 좌표 변경 검사 통과")
print("✅ 600 × 400 RGB 이미지와 PNG 저장 검사 통과")
print("✅ 새 런타임 전체 실행 검사 통과")
print("🎉 WEEK 03 IMAGE MISSION COMPLETE")
```
